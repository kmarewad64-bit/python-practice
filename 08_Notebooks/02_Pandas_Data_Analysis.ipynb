# Pandas Data Analysis

This notebook covers Pandas concepts used in Data Analysis.

```python
# Create a DataFrame

import pandas as pd

data = {
    "Name": ["Komal", "Rahul", "Priya", "Amit", "Sneha"],
    "Age": [21, 22, 20, 23, 21],
    "Marks": [85, 78, 92, 67, 88],
    "Course": ["ENTC", "IT", "ENTC", "Computer", "ENTC"]
}

df = pd.DataFrame(data)

print(df)
```

```python
# View DataFrame Information

print("First 5 Rows:")
print(df.head())

print("\nData Types:")
print(df.dtypes)

print("\nNumber of Rows and Columns:")
print(df.shape)
```

```python
# Selecting Columns

print("Student Names:")
print(df["Name"])

print("\nStudent Marks:")
print(df["Marks"])
```

```python
# Filtering Data

high_marks = df[df["Marks"] >= 80]

print("Students with marks 80 or above:")
print(high_marks)
```

```python
# Sorting Data

sorted_df = df.sort_values("Marks", ascending=False)

print("Students sorted by marks:")
print(sorted_df)
```

```python
# Calculate Average Marks

average_marks = df["Marks"].mean()

print("Average Marks:", average_marks)
```

```python
# Find Highest and Lowest Marks

highest_marks = df["Marks"].max()
lowest_marks = df["Marks"].min()

print("Highest Marks:", highest_marks)
print("Lowest Marks:", lowest_marks)
```

```python
# Add Result Column

df["Result"] = df["Marks"].apply(
    lambda x: "Pass" if x >= 40 else "Fail"
)

print(df)
```

```python
# Count Students by Course

course_count = df["Course"].value_counts()

print("Number of Students in Each Course:")
print(course_count)
```

```python
# Average Marks by Course

course_average = df.groupby("Course")["Marks"].mean()

print("Average Marks by Course:")
print(course_average)
```
