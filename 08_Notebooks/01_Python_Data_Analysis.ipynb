# Python Data Analysis Basics

This notebook covers basic Python concepts used in Data Analysis
 

```python
# Variables and Data Types

name = "Komal"
age = 21
percentage = 85.5
is_student = True

print("Name:", name)
print("Age:", age)
print("Percentage:", percentage)
print("Student:", is_student)

print("\nData Types:")
print(type(name))
print(type(age))
print(type(percentage))
print(type(is_student))
```

```python
# Lists and Dictionaries

skills = ["Python", "Excel", "Power BI", "IoT"]

print("Skills:", skills)
print("First Skill:", skills[0])
print("Number of Skills:", len(skills))

student = {
    "name": "Komal",
    "course": "ENTC Engineering",
    "year": 4
}

print("\nStudent Information:")
print("Name:", student["name"])
print("Course:", student["course"])
print("Year:", student["year"])
```

```python
# Functions

def greet_user(name):
    return "Hello, " + name


def calculate_percentage(marks, total_marks):
    percentage = (marks / total_marks) * 100
    return percentage


print(greet_user("Komal"))

marks = 425
total_marks = 500

result = calculate_percentage(marks, total_marks)

print("Percentage:", result)
```

```python
# Conditional Statements

marks = 85

if marks >= 75:
    print("Grade: Distinction")
elif marks >= 60:
    print("Grade: First Class")
elif marks >= 50:
    print("Grade: Second Class")
elif marks >= 40:
    print("Grade: Pass")
else:
    print("Result: Fail")
```

```python
# Loops

print("Numbers from 1 to 5:")

for i in range(1, 6):
    print(i)

print("\nSquares:")

for i in range(1, 6):
    print(i, "=", i * i)
```

```python
# NumPy Basics

import numpy as np

marks = np.array([78, 85, 92, 67, 88])

print("Marks:", marks)
print("Total Marks:", np.sum(marks))
print("Average Marks:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
```

```python
# NumPy Array Operations

import numpy as np

numbers = np.array([10, 20, 30, 40, 50])

print("Array:", numbers)
print("First Element:", numbers[0])
print("Last Element:", numbers[-1])
print("Array Shape:", numbers.shape)
print("Array Size:", numbers.size)

print("Array + 5:", numbers + 5)
print("Array × 2:", numbers * 2)
```

```python
# Pandas Basics

import pandas as pd

data = {
    "Name": ["Komal", "Rahul", "Priya", "Amit"],
    "Marks": [85, 78, 92, 67],
    "Course": ["ENTC", "IT", "ENTC", "Computer"]
}

df = pd.DataFrame(data)

print("Student Data:")
print(df)
```

```python
# Exploring Pandas Data

print("First 2 Rows:")
print(df.head(2))

print("\nData Information:")
print(df.info())

print("\nMarks Statistics:")
print(df["Marks"].describe())
```

```python
# Filtering Data

high_marks = df[df["Marks"] >= 80]

print("Students with marks 80 or above:")
print(high_marks)
```

```python
# Sorting Data and Adding a New Column

sorted_df = df.sort_values("Marks", ascending=False)

print("Students sorted by marks:")
print(sorted_df)

df["Result"] = df["Marks"].apply(
    lambda x: "Pass" if x >= 40 else "Fail"
)

print("\nData with Result:")
print(df)
```

```python
# Basic Data Visualization

import matplotlib.pyplot as plt

plt.bar(df["Name"], df["Marks"])

plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.title("Student Marks")

plt.show()
```

```python
# Line Chart

plt.plot(df["Name"], df["Marks"], marker="o")

plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.title("Student Marks - Line Chart")

plt.show()
```

```python
# Pie Chart

plt.pie(
    df["Marks"],
    labels=df["Name"],
    autopct="%1.1f%%"
)

plt.title("Marks Distribution")

plt.show()
```

```python
# Histogram

plt.hist(df["Marks"], bins=4)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Marks")

plt.show()
```
